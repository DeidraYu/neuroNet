```python
import os
import subprocess


# Get the current directory of the notebook/script
current_directory = os.getcwd()

# Construct the absolute path to the executable
executable_path = os.path.join(current_directory, '..', '..', 'build', 'src', 'Release', 'neuronet.exe')

# Convert the path to use forward slashes (if needed, for cross-platform support)
executable_path = executable_path.replace('\\', '/')

def neuronet(nEpochs, miniBatchSize, learningRate, layerSizes, seed, verbosity):

    executable_path = "../../build/src/Release/neuronet.exe"

    # Define the command and arguments
    command = [executable_path, str(nEpochs), str(miniBatchSize), str(learningRate), str(layerSizes), str(seed), str(verbosity)]

    # Run the command and capture the output
    process = subprocess.Popen(
        command,
        stdout=subprocess.PIPE,
        stderr=subprocess.PIPE,
        text=True,
        bufsize=1,  # Line-buffered mode
        universal_newlines=True  # Use newline conversion for output
    )

    # Iterate over lines of the output in real-time
    for line in process.stdout:
        print(line, end='')  # Print without adding an extra newline

    # Wait for the process to finish
    process.wait()

    # Print the process return code
    print("Return code:", process.returncode)
```

```python
import neuronetpy as neuronet

# Load and convert the data for our use
mnist_data = neuronet.MnistData()
mnist_data.processMNIST()

# Create an instance of a neuroNet and initialize Weights and Biasses
net = neuronet.Network([784, 30, 10])
net.randomizeWB(-0.0001, 0.0001)

# Create a network trainer
networkTrainer = neuronet.NetworkTrainer()

# Train the network
networkTrainer.trainNet(net, nEpochs=10, learningRate=1.0, miniBatchSize=10, mnistData=mnist_data)

# Evaluate the network
score = networkTrainer.evalNet(net, mnist_data)
print(score)
```
